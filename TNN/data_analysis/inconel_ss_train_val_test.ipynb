```python
train_output_path = '/home/vpatro/TNN_data/inconel_data/input_train_data.npy'
train_input_path = '/home/vpatro/TNN_data/inconel_data/output_train_data.npy'
test_input_path = '/home/vpatro/TNN_data/inconel_data/output_test_data.npy'
test_output_path = '/home/vpatro/TNN_data/inconel_data/input_test_data.npy'


train_output = np.load(train_output_path)
train_input = np.load(train_input_path)
x_test = np.load(test_input_path)
y_test = np.load(test_output_path)

x_train, x_val, y_train, y_val = train_test_split(train_input, 
                                                            train_output,
                                                            test_size=0.2, 
                                                            shuffle=False, 
                                                            random_state=23)


plt.hist(x_train.flatten(), label = 'train', color = 'blue', alpha=0.5, bins=20)

plt.hist(x_val.flatten(), label = 'val', color = 'orange', alpha=0.5, bins=20)

plt.hist(x_test.flatten(), label = 'test', color = 'green', alpha=0.5, bins=20)

plt.title(f'Inconel Emissivity Values - Histogram')
plt.legend()
plt.show()

plt.hist(y_train.flatten(), label = 'train', color = 'blue', alpha=0.5, bins=20)

plt.hist(y_val.flatten(), label = 'val', color = 'orange', alpha=0.5, bins=20)

plt.hist(y_test.flatten(), label = 'test', color = 'green', alpha=0.5, bins=20)
plt.title(f'Inconel Laser Param Values - Histogram')
plt.legend()
plt.show()


```
