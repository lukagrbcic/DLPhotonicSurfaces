```python
train_output_path = '/home/vpatro/TNN_data/airfoil_Re_1_3_data/input_train_data.npy'
train_input_path = '/home/vpatro/TNN_data/airfoil_Re_1_3_data/output_train_data.npy'
test_input_path = '/home/vpatro/TNN_data/airfoil_Re_1_3_data/output_test_data.npy'

train_output = np.load(train_output_path)
train_input = np.load(train_input_path)
x_test = np.load(test_input_path)

x_train, x_val, y_train, y_val = train_test_split(train_input, 
                                                            train_output,
                                                            test_size=0.1, 
                                                            shuffle=False, 
                                                            random_state=23)


counts, edges, _ = plt.hist(x_train.flatten(), label = ' train', color = 'blue', alpha=0.5, bins=10)

for i, count in enumerate(counts):
    print(f"Bin {i+1}: Count = {int(count)}, Range = ({edges[i]:.2f}, {edges[i+1]:.2f})")

counts, edges, _ = plt.hist(x_val.flatten(), label = ' val', color = 'orange', alpha=0.5, bins=10)

for i, count in enumerate(counts):
    print(f"Bin {i+1}: Count = {int(count)}, Range = ({edges[i]:.2f}, {edges[i+1]:.2f})")

counts, edges, _ = plt.hist(x_test.flatten(), label = ' test', color = 'green', alpha=0.5, bins=10)

for i, count in enumerate(counts):
    print(f"Bin {i+1}: Count = {int(count)}, Range = ({edges[i]:.2f}, {edges[i+1]:.2f})")

plt.title(f'Airfoil_Re_1_3_data Values - Histogram')
plt.legend()
plt.show()
```
