```python
import numpy as np
import pandas as pd
import json
```

```python
dataset_name = 'inconel'
hot_start_dataset = 'stainless_steel'


df = pd.DataFrame()

df['Config'] = ['from_scratch', 'hot_start', 'hot_start', 'hot_start', 'hot_start']
df['Hot Start Type'] = ['n/a', 'partial', 'partial', 'full', 'full']
df['Layers Transferred'] = ['n/a', '1', '2', '1', '2']

result_dir = 'forwardDNN_results'
standard_path = f'{result_dir}/standard_{dataset_name}_dataset.json'
tl_1_layer_partial_path = f'{result_dir}/transfer_learning_{dataset_name}_dataset_1_layer_{hot_start_dataset}_PARTIAL_hot_start.json'
tl_2_layer_partial_path = f'{result_dir}/transfer_learning_{dataset_name}_dataset_2_layer_{hot_start_dataset}_PARTIAL_hot_start.json'
tl_1_layer_full_path = f'{result_dir}/transfer_learning_{dataset_name}_dataset_1_layer_{hot_start_dataset}_FULL_hot_start.json'
tl_2_layer_full_path = f'{result_dir}/transfer_learning_{dataset_name}_dataset_2_layer_{hot_start_dataset}_FULL_hot_start.json'

paths = [standard_path, tl_1_layer_partial_path, tl_2_layer_partial_path, tl_1_layer_full_path, tl_2_layer_full_path]

train_losses = []
val_losses = []
test_losses = []
train_std = []
val_std = []
test_std = []
epochs = []
epochs_std = []

for path in paths:

    with open(path, 'r') as f:
        data = json.load(f)

    train_losses.append(data['mean train loss'])
    val_losses.append(data['mean val loss'])
    test_losses.append(data['test loss'])
    train_std.append(data['train loss std'])
    val_std.append(data['val loss std'])
    test_std.append(data['test loss std'])
    epochs.append(data['epochs'])
    epochs_std.append(data['epochs std'])

train_col = [f'{tl:.5f} (+/- {tl_std:.5f})' for tl, tl_std in zip(train_losses, train_std)]
val_col = [f'{vl:.5f} (+/- {vl_std:.5f})' for vl, vl_std in zip(val_losses, val_std)]
test_col = [f'{tl:.5f} (+/- {tl_std:.5f})' for tl, tl_std in zip(test_losses, test_std)]
epoch_col = [f'{e} (+/- {e_std:.2f})' for e, e_std in zip(epochs, epochs_std)]

df['Train RMSE'] = train_col
df['Val RMSE'] = val_col
df['Test RMSE'] = test_col
df['Epochs'] = epoch_col

df = df.set_index('Config')

df
```
